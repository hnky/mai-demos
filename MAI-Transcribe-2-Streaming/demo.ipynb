# MAI-Transcribe-2-Streaming: A Bit of AI goes live

Henk is opening a live episode of **A Bit of AI**. Can the audience read his
welcome while he is still speaking?

This standalone demo follows **prepare -> speak -> watch evolving captions ->
stop -> review the final transcript**. It uses MAI-Transcribe-2-Streaming,
not repeated file uploads or the file-transcription API.

**Experimental:** browser microphone capture and the live Azure path still
need end-to-end verification in the presentation environment. Rendering the
controls does not prove that the stream works. The model itself is a public
preview without an SLA and is not recommended for production.

Follow the [setup instructions](demo-instructions.md) first.
[Extended model information](information.md) covers languages, regions,
protocol, costs, and limitations.

**Cost and privacy:** running these cells only renders controls. Clicking
**Start captions** requests the browser microphone and starts a paid Azure
session. Use permitted, non-sensitive speech. No microphone audio file is saved;
credentials stay in Python. No other notebook or sample recording is needed.

## 1. Prepare the live desk

Use this folder's Python environment and `.env`, or Codespaces secrets:
`AZURE_MAI_ENDPOINT`, `AZURE_MAI_DEPLOYMENT_NAME`, and optional `AZURE_MAI_API_KEY`.
The endpoint is a Foundry `services.ai.azure.com` resource root; the model
selection is your deployed streaming model's name.

A configured key is used directly. Without one, sign in with `az login`
(`az login --use-device-code` in Codespaces) and give that identity resource access.
Nothing reads the file demo's `SPEECH_*` or the Voice demo's settings.

Use a browser on HTTPS/localhost with microphone and widget support. Embedded
notebooks may block capture; the instructions include a full-tab JupyterLab
route for Codespaces. Permissions still need to be verified on the actual host.
Do not paste credentials into a cell.

## 3. Finish the welcome and review the text

Click **Stop**. Microphone capture stops, the final buffered audio is sent,
and the client asks Azure to finalize. Wait for **Captions finalized**:
a commit acknowledgement alone does not mean transcription has finished.

After successful finalization, `captions.transcript` holds the final text in
kernel memory. After an error it may be incomplete. Review names, numbers,
and corrections before sharing it. There is no automatic export or summary.

The demo caps capture at five minutes and stops when you leave the browser tab.
Closing the widget cancels its session; lost connections and overflowing buffers
produce visible errors. A new Start is a new paid session, not an automatic retry.

The expected result is a captioned welcome, not a promise of a fixed latency.
If the live path fails, keep the displayed error and host/browser details for
diagnosis, without sharing credentials or private speech.
